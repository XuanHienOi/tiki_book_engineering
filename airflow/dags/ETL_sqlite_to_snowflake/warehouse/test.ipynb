```python
import os
import snowflake.connector
from dotenv import load_dotenv

ENV_PATH = "/home/phinguyen/projects/tiki_book_engineering/airflow/.env"

load_dotenv(ENV_PATH)
```

```python
account_identifier = os.getenv("SNOWFLAKE_ACCOUNT_IDENTIFIER")
username = os.getenv("SNOWFLAKE_USERNAME")
password = os.getenv("SNOWFLAKE_PASSWORD")
warehouse = os.getenv("SNOWFLAKE_WAREHOUSE")
database = os.getenv("SNOWFLAKE_DATABASE")
schema = os.getenv("SNOWFLAKE_SCHEMA")

print(f"Account Identifier: {account_identifier}")
print(f"Username: {username}")
print(f"Password: {password}")
print(f"Warehouse: {warehouse}")
print(f"Database: {database}")
print(f"Schema: {schema}")

def connect_to_snowflake(
    account_identifier=account_identifier,
    username=username,
    password=password
):
    try:
        conn = snowflake.connector.connect(
            user=username,
            password=password,
            account=account_identifier
        )
        print("Connection to Snowflake was successful!")

        cursor = conn.cursor()
        cursor.execute("SELECT current_version()")
        version = cursor.fetchone()
        print(f"Snowflake version: {version[0]}")
        cursor.close()
    except Exception as e:
        print(f"Error connecting to Snowflake: {e}")
    finally:
        if conn:
            conn.close()

connect_to_snowflake()
```
